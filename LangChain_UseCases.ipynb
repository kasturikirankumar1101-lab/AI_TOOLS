```python
"""
================================================================================
COMPLETE FUNCTIONALITY OVERVIEW
================================================================================
This script demonstrates how to generate and compare text embeddings using
OpenAI's embedding model via LangChain.

Step-by-step functionality:

1. Environment Setup:
   - Loads environment variables using dotenv.
   - Securely retrieves the OpenAI API key from Google Colab secrets.
   - Sets the API key as an environment variable so LangChain can use it.

2. Embedding Model Initialization:
   - Uses OpenAI's 'text-embedding-3-large' model.
   - Configures the embedding vector size to 32 dimensions (compressed vector).

3. Embedding Generation:
   - Converts input text into numerical vector representations (embeddings).
   - Two semantically similar sentences are embedded:
        a. "Delhi is the capital of India"
        b. "India's capital is Delhi"

4. Output:
   - Prints both embedding vectors.
   - Compares the embeddings using string equality.

5. Key Insight:
   - Even though both sentences have the same meaning,
     their embeddings are NOT exactly identical.
   - This is because embeddings capture semantic meaning in a continuous space,
     not exact textual equivalence.

6. Real-world Note:
   - Instead of direct equality (==), similarity should be measured using:
        - Cosine similarity (recommended)
        - Euclidean distance

Use Cases:
   - Semantic search
   - RAG (Retrieval-Augmented Generation)
   - Clustering similar documents
   - Recommendation systems
================================================================================
"""

# Import the OpenAIEmbeddings class from langchain_openai
# This is used to generate vector embeddings for text using OpenAI models
from langchain_openai import OpenAIEmbeddings

# Import load_dotenv to load environment variables from a .env file
from dotenv import load_dotenv

# Import userdata module from Google Colab
# This allows secure access to stored secrets (like API keys) in Colab
from google.colab import userdata

# Import os module to interact with environment variables
import os


# Load environment variables from a .env file into the environment
# This ensures any locally stored variables are available in runtime
load_dotenv()

# Retrieve the OpenAI API key securely from Google Colab secrets
# 'OPENAI_API_KEY' should be stored in Colab's userdata
openai_api_key = userdata.get('OPENAI_API_KEY')


# Set the OpenAI API key as an environment variable
# LangChain automatically reads this variable when making API calls
os.environ["OPENAI_API_KEY"] = openai_api_key

# Print confirmation message to indicate API key is successfully loaded
print("OpenAI API key loaded from Colab secrets.")


# Initialize the OpenAIEmbeddings model
# model='text-embedding-3-large' specifies the embedding model to use
# dimensions=32 reduces the embedding vector size (default is larger)
embedding = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=32)


# Generate embedding vector for the given query sentence
# embed_query converts text into a numerical vector representation
result = embedding.embed_query("Delhi is the capital of India")

# Print the embedding vector as a string
# This will output a list of numerical values representing the text
print(str(result))


# Generate embedding vector for a semantically similar sentence
result1 = embedding.embed_query("India's capital is Delhi")

# Compare both embedding vectors by converting them to strings
# This checks if both embeddings are exactly identical (unlikely in practice)
# Even semantically similar sentences usually produce slightly different vectors
print(str(result1) == str(result))
```

```python
"""
================================================================================
COMPLETE FUNCTIONALITY OVERVIEW
================================================================================
This script demonstrates how to use a Hugging Face hosted Large Language Model
(LLM) via LangChain to generate a conversational response.

Step-by-step functionality:

1. Environment Setup:
   - Loads environment variables using dotenv.
   - Securely retrieves the Hugging Face API token from Google Colab secrets.
   - Sets the API token as an environment variable so LangChain can use it.

2. Model Initialization (Hugging Face Endpoint):
   - Connects to a hosted Hugging Face model using HuggingFaceEndpoint.
   - Uses the model: "igorls/gemma-4-12B-it-heretic-GGUF".
   - Configures the task as "text-generation".

3. Chat Wrapper:
   - Wraps the Hugging Face LLM with ChatHuggingFace.
   - This enables chat-style interaction (message-based interface).

4. Model Invocation:
   - Sends a user query: "What is the capital of India".
   - The model processes the input and generates a response.

5. Output:
   - Extracts and prints the generated response content.

6. Key Insight:
   - HuggingFaceEndpoint allows you to use open-source or hosted models
     similar to OpenAI APIs.
   - ChatHuggingFace provides a standardized interface for conversational use.

7. Real-world Use Cases:
   - Chatbots and virtual assistants
   - AI copilots
   - Content generation
   - Question-answering systems
   - Multi-model orchestration pipelines

================================================================================
"""

# Import ChatHuggingFace and HuggingFaceEndpoint from langchain_huggingface
# ChatHuggingFace provides a chat interface wrapper for Hugging Face models
# HuggingFaceEndpoint allows calling hosted Hugging Face models via API
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

# Import load_dotenv to load environment variables from a .env file
from dotenv import load_dotenv

# Import userdata module from Google Colab
# This allows secure access to stored secrets (like API tokens) in Colab
from google.colab import userdata

# Import os module to manage environment variables
import os


# Load environment variables from .env file
# Ensures any locally stored variables are available in runtime
load_dotenv()


# Retrieve the Hugging Face API key from Colab secrets
# Make sure you've added your HUGGINGFACEHUB_ACCESS_TOKEN to Colab's secrets manager
HF_TOKEN = userdata.get('HUGGINGFACEHUB_ACCESS_TOKEN')


# Set the environment variable for LangChain to pick it up
# This allows LangChain to authenticate with Hugging Face services
os.environ["HF_TOKEN"] = HF_TOKEN


# Initialize the Hugging Face model endpoint
# repo_id specifies which model to use from Hugging Face Hub
# task defines the type of operation (text-generation in this case)
llm = HuggingFaceEndpoint(
    repo_id = "deepseek-ai/DeepSeek-V4-Flash",
    task = "text-generation"
)


# Wrap the LLM with ChatHuggingFace
# This converts the raw LLM into a chat-compatible interface
model = ChatHuggingFace(llm=llm)


# Invoke the model with a user query
# The model processes the input and generates a response
result = model.invoke("What is the capital of India")


# Print only the content of the response
# result.content contains the generated text output from the model
print(result.content)
```
