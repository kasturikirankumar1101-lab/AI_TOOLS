```python
# Stemming
from nltk.stem import PorterStemmer

porter = PorterStemmer()
words = ['Cats','Catering', 'bowling', 'friends', 'routers', 'plyaers', 'playing', 'starting', 'starts', 'started']

for word in words:
  print(word,"--->",porter.stem(word))
```

```python
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer(language='english')
words = ['Cats','Catering', 'bowling', 'friends', 'routers', 'plyaers', 'playing', 'starting', 'starts', 'started']

for word in words:
  print(word,"--->",snowball.stem(word))
```

```python
# Combining both Tokenizer, stop_word, stemming

text = '''Natural Language Processing Is evolving rapidly in today’s data-driven world, and many organizations are investing heavily in NLP-based solutions to extract meaningful insights from unstructured text data. In 2025, companies are not only analyzing customer feedback but also building intelligent chatbots, automated summarization systems, and real-time translation tools that operate at scale. However, before applying advanced models, it is absolutely essential to perform proper text preprocessing steps such as tokenization, stop word removal, normalization, and stemming or lemmatization. For example, words like "running", "runs", and "ran" should ideally be reduced to a common base form during stemming, while common stop words such as "is", "the", "and", and "in" are typically removed to reduce noise in the dataset. Additionally, handling punctuation marks, special characters (like @, #, $, %), and numerical values such as 123 or 45.67 can significantly impact the quality of the processed text. Consider a scenario where a user writes, "I am LOVING this product!!! It’s absolutely amazing, and I’ve never seen anything better." In this case, tokenization should correctly split the sentence into meaningful tokens, case folding should convert "LOVING" into "loving", and stemming might reduce it to "love". Furthermore, repeated words, elongated expressions like "soooo good", and contractions such as "don’t" or "I’ve" must be handled carefully depending on the use case. While preprocessing may seem simple, it plays a crucial role in ensuring that downstream machine learning models perform efficiently and accurately across diverse datasets.'''

```

```python
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
print(doc)
```

```python
tokens = [token for token in doc]
print(tokens)
```

```python
# stop_words now
stop_word_doc = [token for token in doc if token.text.lower() not in nlp.Defaults.stop_words]
print(stop_word_doc)
```

```python
nlp.vocab["'s"].is_stop
nlp.Defaults.stop_words.remove("natural")
```

```python
# Stemming now
from nltk.stem import PorterStemmer
porter = PorterStemmer()

for words in stop_word_doc:

  print(words.text,"----->", porter.stem(words.text.lower()))
```

```python
# Stemming now
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english')


for words in stop_word_doc:

  print(words.text,"----->", snowball.stem(words.text.lower()))
```
