```python
#This program would generate an Image whichever you like it.

import torch
import gradio as gr
from functools import lru_cache
from google.colab import userdata
from huggingface_hub import login
from diffusers import AutoPipelineForText2Image

# -----------------------------
# 🔐 Login to Hugging Face
# -----------------------------
hf_token = userdata.get('HF_TOKEN')

if hf_token and hf_token.startswith("hf_"):
    print("✅ HF key looks good")
    login(hf_token)
else:
    print("❌ HF key missing")

# -----------------------------
# ⚙️ Model Loader (cached)
# -----------------------------
@lru_cache(maxsize=1)
def get_model_pipeline(model_name):
    print(f"🚀 Loading {model_name} (only once)")

    device = "cuda" if torch.cuda.is_available() else "cpu"

    pipe = AutoPipelineForText2Image.from_pretrained(
        model_name,
        torch_dtype=torch.float16 if device == "cuda" else torch.float32
    ).to(device)

    return pipe


# -----------------------------
# 🧠 Main Function
# -----------------------------
def image_generate(text):
    if not text:
        return None
        print("It is giving error")

    pipeline = get_model_pipeline("playgroundai/playground-v2.5-1024px-aesthetic")

    image = pipeline(text,num_inference_steps=25,
        guidance_scale=7.5).images[0]
    return image


# -----------------------------
# 🎨 Gradio UI
# -----------------------------
with gr.Blocks() as demo:
    gr.Markdown("## 🎨 AI Image Generator (Stable Diffusion)")

    text_input = gr.Textbox(lines=3, label="Enter Prompt")

    output = gr.Image(label="Generated Image")

    run_btn = gr.Button("Generate Image")

    run_btn.click(
        fn=image_generate,
        inputs=text_input,
        outputs=output
    )

# -----------------------------
# 🚀 Launch
# -----------------------------
demo.launch(debug=True)
```
